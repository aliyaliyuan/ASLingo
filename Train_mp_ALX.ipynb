```python
import mediapipe as mp
import cv2
import os
import csv

mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=True, max_num_hands=1)

# Path to your training data
DATA_PATH = "/content/drive/MyDrive/ALX_Train"

# Open CSV file to save landmarks
with open('/content/drive/MyDrive/ALX_Train/landmark_data.csv', 'w', newline='') as f:
    writer = csv.writer(f)

    # Write header
    header = ['label']
    for i in range(21):
        header.extend([f'x{i}', f'y{i}', f'z{i}'])
    writer.writerow(header)

    # Loop through each letter folder
    for letter in ['A', 'L', 'X', 'None']:
        folder_path = os.path.join(DATA_PATH, letter)

        for img_file in os.listdir(folder_path):
            # Skip hidden files
            if img_file.startswith('.'):
                continue

            img_path = os.path.join(folder_path, img_file)
            image = cv2.imread(img_path)

            if image is None:
                continue

            # Convert BGR to RGB
            image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

            # Process image
            results = hands.process(image_rgb)

            if results.multi_hand_landmarks:
                for hand_landmarks in results.multi_hand_landmarks:
                    # Extract coordinates
                    row = [letter]
                    for landmark in hand_landmarks.landmark:
                        row.extend([landmark.x, landmark.y, landmark.z])
                    writer.writerow(row)
            else:
                print(f"No hand detected in {img_path}")

hands.close()
print("Landmark extraction complete! Check landmark_data.csv")
```

```python
#Train classifier
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# Load the data
df = pd.read_csv('/content/drive/MyDrive/ALX_Train/landmark_data.csv')

# Check how many rows have NaN
print(f"Total rows: {len(df)}")
print(f"Rows with NaN: {df.isnull().any(axis=1).sum()}")

# Drop rows with any NaN values
df = df.dropna()
print(f"Rows after dropping NaN: {len(df)}")

# Separate features (X) and labels (y)
X = df.drop('label', axis=1)
y = df['label']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train
clf = RandomForestClassifier(n_estimators=100)
clf.fit(X_train, y_train)

# Test
accuracy = clf.score(X_test, y_test)
print(f"Accuracy: {accuracy}")

#Save model with joblib
joblib.dump(clf, '/content/drive/MyDrive/ASLingo/ALX_model.joblib')
```

```python


# After training your model, test on a new image
import cv2
import mediapipe as mp

# Reinitialize MediaPipe Hands
mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=True, max_num_hands=1)

# Load new test image
test_image = cv2.imread('/content/drive/MyDrive/ASLingo/A.jpg')
test_image_rgb = cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB)

# Extract landmarks
results = hands.process(test_image_rgb)
if results.multi_hand_landmarks:
    landmarks = []
    for landmark in results.multi_hand_landmarks[0].landmark:
        landmarks.extend([landmark.x, landmark.y, landmark.z])

    # Predict
    prediction = clf.predict([landmarks])
    print(f"Predicted: {prediction[0]}")
```
